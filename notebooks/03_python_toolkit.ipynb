### 1. Tracking bugs
**Example A:** Profit is wrong because unit cost was subtracted only once. Record input, expected result, actual result, and reproduction steps.

**Example B:** Duplicate orders inflate revenue. Record the affected order IDs and agree on a deduplication rule.

A simple issue template: **Title → Steps → Expected → Actual → Evidence → Status**.


````python
units, price, cost = 10, 100, 60
wrong_profit = units * price - cost
correct_profit = units * (price - cost)
print('Incorrect:', wrong_profit, '| Correct:', correct_profit)
````

### 3. Developer documentation with pdoc
**Example A:** document a profit function.

**Example B:** document a small analysis module containing several functions.

Save documented functions in `business_metrics.py`, then run in a terminal:

```bash

```

This produces API reference documentation; business assumptions still need a written explanation. [Official pdoc documentation](https://pdoc.dev/).


````python
def gross_profit(units, price, unit_cost):
    """Return gross profit in INR, excluding overheads and taxes."""
    return units * (price - unit_cost)

def revenue(units, price):
    """Return sales revenue in INR before tax adjustments."""
    return units * price

print(gross_profit.__doc__)
print(revenue.__doc__)
````

### 7. Testing Python applications: PyUnit / unittest
Tests check known expectations. **Example A:** profit on a normal sale. **Example B:** no units sold gives no revenue. The suite runs inside the notebook without calling `unittest.main()`.


````python
import unittest

class TestBusinessMetrics(unittest.TestCase):
    def test_gross_profit(self):
        self.assertEqual(gross_profit(10, 100, 60), 400)

    def test_zero_units_revenue(self):
        self.assertEqual(revenue(0, 100), 0)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestBusinessMetrics)
result = unittest.TextTestRunner(verbosity=2).run(suite)
assert result.wasSuccessful()
````
