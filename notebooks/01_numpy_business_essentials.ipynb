## 1. A quick Python refresher
A **list** stores a sequence. A **dictionary** maps keys to values. A **library** offers reusable functionality; a **package** organises importable modules. NumPy and Pandas are libraries distributed as Python packages.


```python
prices = [100, 200, 300]
branch_targets = {'Lucknow': 50000, 'Delhi': 70000}
print('First price:', prices[0])
print('Lucknow target:', branch_targets['Lucknow'])
```

## 2. Create and inspect an array
Sales values below are in INR. An array supports calculations across all its values.


```python
import numpy as np
monthly_sales = np.array([42000, 48000, 51000, 45000, 60000, 54000])
print(monthly_sales)
print('Shape:', monthly_sales.shape)
print('Dimensions:', monthly_sales.ndim)
print('Data type:', monthly_sales.dtype)
```

## 3. Select a period
Python indexing starts at zero. Slices include the start and exclude the stop.


```python
print('First month:', monthly_sales[0])
print('First quarter:', monthly_sales[:3])
print('Last two months:', monthly_sales[-2:])
```

## 4. Vectorised calculations
**Example A:** calculate revenue for three products.

**Example B:** model a 10% price increase, assuming units sold stay constant. This is a scenario, not a demand forecast.


```python
units = np.array([20, 15, 10])
prices = np.array([500, 800, 1200])
revenue = units * prices
print('Product revenue:', revenue)
print('Total revenue:', revenue.sum())
new_prices = prices * 1.10
print('Revenue with higher prices:', units * new_prices)
```

## 5. Summarise performance
Mean is the average; median is the middle value. Standard deviation describes spread. Here `std()` uses population standard deviation because we describe these six months.


```python
print('Total:', monthly_sales.sum())
print('Mean:', monthly_sales.mean())
print('Median:', np.median(monthly_sales))
print('Minimum:', monthly_sales.min())
print('Maximum:', monthly_sales.max())
print('Standard deviation:', monthly_sales.std().round(2))
```

## 6. Filter against a target
A condition produces a Boolean mask: one True/False value per item.


```python
target = 50000
print('Met target:', monthly_sales >= target)
print('Sales below target:', monthly_sales[monthly_sales < target])
print('Months meeting target:', np.count_nonzero(monthly_sales >= target))
```

## 7. Compare branches with a matrix
Rows represent branches; columns represent months. `axis=1` totals each row. `axis=0` totals each column.


### Sample matrix: branches × months

A one-dimensional array is one line of values. A two-dimensional array is a matrix: rows can represent branches while columns represent months. NumPy uses the `axis` argument to choose which direction to calculate.

![Sample NumPy array and matrix](https://raw.githubusercontent.com/zaid-kamil/Data-analytics-using-numpy-and-python-at-Amity-Lucknow/main/assets/diagrams/numpy-array-matrix.svg)


```python
branch_names = np.array(['Lucknow', 'Delhi', 'Kanpur'])
branch_sales = np.array([[42000, 48000, 51000],
                         [55000, 50000, 60000],
                         [38000, 41000, 44000]])
print('Branch totals:', branch_sales.sum(axis=1))
print('Monthly totals:', branch_sales.sum(axis=0))
print('Top branch:', branch_names[branch_sales.sum(axis=1).argmax()])
```

## 8. Visualise the monthly sales

A chart makes the array easier to scan. The red line is the target; green bars reached it and blue bars need another cup of coffee.

![Monthly sales bar chart with target](../assets/diagrams/numpy-sales-chart.svg)

The code below builds the same chart directly from NumPy arrays.


```python
import matplotlib.pyplot as plt

months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'])
bar_colours = np.where(monthly_sales >= target, '#22c55e', '#93c5fd')

plt.figure(figsize=(8, 4))
plt.bar(months, monthly_sales, color=bar_colours)
plt.axhline(target, color='#e11d48', linestyle='--', label='Target: INR 50,000')
plt.title('Monthly Sales versus Target')
plt.ylabel('Sales (INR)')
plt.legend()
plt.tight_layout()
plt.show()
```
